# Day 5 - Pandas: Basic Dataset Analysis

This notebook demonstrates the use of Pandas for loading, inspecting, filtering, and analyzing a student dataset.

```python
import pandas as pd
```

## 1. Loading the Dataset

The `read_csv()` function is used to load the CSV dataset into a Pandas DataFrame.

```python
df = pd.read_csv("student_data.csv")
```

## 2. Viewing the Dataset

The `head()` function displays the first five rows of the dataset, while `tail()` displays the last five rows.

```python
print("First 5 rows:") 
display(df.head()) #df.head(n) returns first n rows

print("Last 5 rows:")
display(df.tail()) #df.tail(n) returns last n rows
```

## 3. Dataset Dimensions

The `shape` attribute returns the number of rows and columns present in the DataFrame.

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 4. Dataset Information

The `info()` function provides information about the DataFrame, including column names, data types, and the number of non-null values.

```python
df.info()
```


## 5. Checking Missing Values

The `isnull()` function identifies missing values, and `sum()` counts their number.
`isnull().sum()` counts the missing values in each column.

```python
df.isnull().sum()
```

## 6. Filtering Data

Filtering allows us to select specific rows from the DataFrame based on conditions.

Different filtering conditions are used below to select students based on their marks, attendance, and department.

```python
print("Students with Python marks > 80:")
display(df[df["Python_Marks"] > 80])

print("Students with attendance > 85:")
display(df[df["Attendance"] > 85])

print("AIML students:")
display(df[df["Department"] == "AIML"])
```

### Filtering with Multiple Conditions

The `&` operator is used when both conditions must be true, while the `|` operator is used when either condition can be true.

```python
print("Students with Python > 80 AND ML > 80:")
display(df[(df["Python_Marks"] > 80) & (df["ML_Marks"] > 80)])
```

```python
print("Students with Python > 85 OR ML > 90:")
display(df[(df["Python_Marks"] > 85) | (df["ML_Marks"] > 90)])
```

## 7. Summary Statistics

The `describe()` function provides statistical information for the numerical columns, including count, mean, standard deviation, minimum, maximum, and quartile values.

```python
print("Summary statistics:")
display(df.describe())
```

### Individual Column Statistics

Specific statistical measures can also be calculated for individual columns using functions such as `mean()`, `max()`, and `min()`.

```python
print("Python marks statistics:")
print(df["Python_Marks"].describe())

print("\nAverage Python marks:", df["Python_Marks"].mean())
print("Average ML marks:", df["ML_Marks"].mean())

print("\nHighest Python marks:", df["Python_Marks"].max())
print("Highest ML marks:", df["ML_Marks"].max())

print("\nLowest Python marks:", df["Python_Marks"].min())
print("Lowest ML marks:", df["ML_Marks"].min())
```

## 8. Conclusion

The dataset was successfully loaded and analyzed using Pandas. The analysis included viewing the dataset, checking its dimensions and information, identifying missing values, filtering records using different conditions, and generating summary statistics.
