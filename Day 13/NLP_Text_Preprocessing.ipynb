
# Day 13: Natural Language Processing (NLP)

Natural Language Processing (NLP) is a branch of Artificial Intelligence that enables computers to process and analyze human language. Text preprocessing prepares raw text for analysis, while TF-IDF converts text into numerical features.

## Dataset
The dataset is stored in `nlp.csv` and contains three columns:
- `id`: Unique identifier for each review.
- `review`: Text of the movie review.
- `result`: Sentiment label, where `p` represents positive and `n` represents negative.

## Objectives
1. Load and inspect the dataset.
2. Convert text to lowercase.
3. Tokenize the reviews.
4. Remove stopwords and punctuation.
5. Convert the cleaned text into TF-IDF features.
6. Inspect the resulting feature matrix.



## 1. Importing Libraries and Loading the Dataset

Pandas handles the dataset, NLTK supports text preprocessing, and Scikit-learn converts the cleaned reviews into numerical TF-IDF features.


```python

import pandas as pd
import nltk
from nltk.tokenize import wordpunct_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download("stopwords")

df = pd.read_csv("nlp.csv")

print("Dataset shape:", df.shape)
display(df.head())
```


## 2. Text Preprocessing

Text preprocessing converts raw text into a consistent format for analysis.

- **Lowercasing:** Converts all text to lowercase.
- **Tokenization:** Splits text into individual tokens (words and punctuation).
- **Stopword removal:** Removes common words that provide limited information, while retaining negation words such as "not".
- **Punctuation removal:** Keeps alphabetic tokens for this exercise.


```python

# Convert reviews to lowercase
df["lowercase"] = df["review"].str.lower()

# Split reviews into tokens
df["tokens"] = df["lowercase"].apply(wordpunct_tokenize)

# Get English stopwords
stop_words = set(stopwords.words("english"))

# Keep negation words
stop_words.discard("not")
stop_words.discard("no")
stop_words.discard("nor")

# Remove stopwords and punctuation
cleaned_reviews = []

for tokens in df["tokens"]:
    cleaned_words = []

    for word in tokens:
        if word.isalpha() and word not in stop_words:
            cleaned_words.append(word)

    cleaned_reviews.append(cleaned_words)

df["cleaned_tokens"] = cleaned_reviews

# Join the cleaned words into sentences
df["processed_text"] = df["cleaned_tokens"].apply(" ".join)

display(df[["review", "processed_text", "result"]].head())
```


## 3. TF-IDF Vectorization

TF-IDF (Term Frequency–Inverse Document Frequency) converts text into numerical features. It gives higher importance to terms that are frequent in a particular review but less common across the dataset.


```python

# Create the TF-IDF vectorizer
vectorizer = TfidfVectorizer(
    tokenizer=str.split,
    preprocessor=None,
    token_pattern=None,
    lowercase=False
)

# Convert processed reviews into numerical features
tf_matrix = vectorizer.fit_transform(df["processed_text"])

# Get the words used as features
words = vectorizer.get_feature_names_out()

# Display the TF-IDF values
tf_df = pd.DataFrame(
    tf_matrix.toarray(),
    columns=words,
    index=df["id"]
)

display(tf_df.round(3))

print("Number of reviews:", tf_matrix.shape[0])
print("Number of unique words:", tf_matrix.shape[1])
```


## 4. Conclusion

The movie reviews were processed using lowercasing, tokenization, stopword removal, and punctuation filtering. TF-IDF converted the cleaned reviews into numerical features that can be used for further machine learning tasks, such as sentiment classification.

```python

print("Total reviews:", len(df))
print("Missing reviews:", df["review"].isna().sum())
print("Processed reviews:", df["processed_text"].notna().sum())
print("TF-IDF output shape:", tf_matrix.shape)

display(df[["id", "processed_text", "result"]])
```
